```python
# data 
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y) #line chart 
plt.show()  #display chart 
```

```python
plt.plot(x, y, color='y', marker='o', linestyle='--', linewidth=2, markersize=10)
plt.show()
```

```python
plt.figure(figsize=(4,3)) #figure size inch 
plt.plot(x,y, color = 'blue',marker='o', linestyle='--', linewidth = 2, markersize = 12)
plt.show()
```

```python
plt.figure(figsize=(4,3))  #figure size 
plt.plot(x,y,color = 'y',marker='o', linestyle='--',linewidth=2, markersize = 10)
plt.title('PNINFOSYS')
plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [10,20,25,35,40]
y2 = [20,30,35,45,55]

plt.plot(x,y1, label = 'Sales 2022')  #plotting y1 data 
plt.plot(x,y2, label = 'Sales 2023')  #plotting y2 data 

plt.title('PN Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.legend()  #show legends
plt.show()
```

```python
months = [1,2,3,4,5]
sales = [10,20,25,35,40]
profit = [5,10,15,20,30]

plt.plot(months,sales,label = 'Sales')
plt.plot(months,profit, label = 'Profit')
plt.legend()
plt.show()
```

```python
#line color  style 

plt.plot(
    months,
    sales,
    color = 'red',
    linestyle ='-.',
    linewidth=3
)

plt.show()
#'-' solid line 
#'--' Dashed line
#':' Dotted line
#'-.' Dash dot line
    
```

```python
#Marker use karna 
plt.plot(
    months,
    sales,
    marker = '^'
)
plt.show()
#'o' circle 
#'s' square 
#'^' triangle 
# '*' star
```

```python
#Title & labels
plt.plot(months,sales)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.show()
#professional reporting ke liye zaruri 
```

```python
#Figure size Control 
plt.figure(figsize=(3,5))
plt.plot(months,sales)
plt.show()
 #width = 10 
#height = 5
```

```python
# Data points  Annotate karna 
plt.plot(months, sales, marker='o')
for x,y in zip(months,sales):
     plt.text(x,y,str(y))
    
plt.show()

#Har point ke upar value dikhegi
```

```python
# Pandas Dataframe ke sath 
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Month':[1,2,3,4,5],
    'Sales':[10,20,15,5,10]
})
plt.plot(df['Month'],df['Sales'])
plt.show()

#real data Analytics mai isi tarah use hota hai 
```

```python
#Trend Analysis Example 
months = ['Jan','Feb','Mar','Apr','May']
sales = [10000,15000,18000,25000,350000]

plt.plot(months,sales,marker ='o')
plt.title('Sales Growth Trend')
plt.show()

#Business Analyst isi type ka graph banate h 
```

```python
#professional Advanced Example 
months = ['Jan','Feb','Mar','Apr','May']
sales = [10000,15000,18000,25000,350000]

profit = [3000,5000,7000,10000,15000]
plt.figure(figsize = (10,5))
plt.plot(months, sales,
         marker='o',
         linewidth=3,
         label = 'Sales')

plt.plot(months,profit,
         marker='s',
         linewidth = 3,
         label = 'Profit')

plt.title('Sales v/s Profit Analysis')
plt.xlabel('Months')
plt.ylabel('Amount')
plt.grid(True)
plt.legend()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,25,35,45]
plt.plot(x,y,marker='o')

plt.grid(
    True,
    color='gray',
    linestyle='--',
    linewidth=1,
    alpha=0.6 #axis = 'y'x linestyle='--' color ='red' linewidth = 2 alpha 0.5 = invisible = 1 fully visible 
)
plt.show()
```

```python
#Grid Add karna 
plt.plot(months,sales)
plt.grid(True)  # axis= 'y'x linestyle ='--' color='red' linewidth = 2 alpha =0.5 0 = invisible 1 = fully visible 

plt.show()
# grid graph ko read karne aur values compare karne ko easy banati hai 
```
