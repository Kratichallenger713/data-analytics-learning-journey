```python
import matplotlib.pyplot as plt
# Scatter Plot do numerical variables ke beech relationship dikhata hai.

# Example:

# Hours Studied vs Marks
# Height vs Weight
# Experience vs Salary

# Har data point ek dot (•) ke roop me dikhaya jata hai.
```

```python
hours = [1,2,3,4,5]
marks = [20,40,50,70,10]

plt.scatter(hours,marks)
plt.title('Hours Studied Vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
plt.scatter(
    hours,
    marks,
    color='red'
)
plt.show()
```

```python
plt.scatter(
    hours,
    marks,
    marker='*'
)
```

```python
plt.scatter(
    hours,
    marks,
    s=200  # s = size 
)
```

```python
# | Marker | Shape    |
# | ------ | -------- |
# | 'o'    | Circle   |
# | '*'    | Star     |
# | 's'    | Square   |
# | '^'    | Triangle |
# | 'x'    | Cross    |
```

```python
hours = [1,2,3,4,5]
marks=[20,40,50,70,90]

plt.scatter(
    hours,
    marks,
    color='blue',
    s =100
)
plt.title('Hours Studied V/S Marks')
plt.xlabel(' Study Hours')
plt.ylabel('Marks')
plt.grid()
plt.show()
```

```python
height = [150,155,160,165,170,175]
weight = [45,50,55,60,65,70]

plt.scatter(height,weight)
plt.title('Height v/s Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.show()
         
```

```python
# Scatter Plot Kya Hai?

# Scatter Plot ek graph hai jo do numerical variables ke beech relationship ko points (dots) ke roop me dikhata hai.

# Scatter Plot Kab Use Karte Hain?
# Correlation dekhne ke liye
# Trends identify karne ke liye
# Outliers find karne ke liye
    
```
