```python
import pandas as pd 
```

```python
data = {
    'Age':[20,25,30,35,40],
    'Salary':[20000,25000,30000,35000,40000]
}
df= pd.DataFrame(data)
print(df.describe())
```

```python
# 1️⃣ count

# Batata hai kitni values available hain (NaN ko count nahi karta).

# Age
# 20
# 25
# 30
# 35
# 40

# Total values = 5

# Output

# count = 5
```

```python
# 2️⃣ mean (Average)

# Formula

# Mean=
# Total values
# Sum of all values
# 	​


# Example

# 20+25+30+35+40=150
# 150/5=30

# Output

# mean = 30
```

```python
# 3️⃣ std (Standard Deviation)

# Ye batata hai data average se kitna spread (faila hua) hai.

# Std kam → Data ek jaisa hai.
# Std zyada → Data bahut spread hai.

# Example

# Students ke marks

# 50
# 51
# 50
# 49
# 50

# Std = Bahut kam

# Dusra example

# 10
# 90
# 30
# 70
# 50

# Std = Bahut zyada
```

```python
# 6️⃣ 25%

# Isko First Quartile (Q1) bolte hain.

# Matlab

# 👉 25% data is value ke niche hai.

# Example

# 10
# 20
# 30
# 40
# 50
# 60
# 70
# 80

# 25% = 27.5 (approx.)
```

```python
# 7️⃣ 50%

# Isko Median bolte hain.

# Ye beech wali value hoti hai.

# Example

# 10
# 20
# 30
# 40
# 50

# Middle value

# 30

# Output

# 50% = 30
```
