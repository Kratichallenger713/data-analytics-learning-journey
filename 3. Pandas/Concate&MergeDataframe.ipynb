```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'Name':['Ram','Ravi'],
    'Age':[25,30]
})

df2 = pd.DataFrame({
    'Name':['Amit','Sita'],
    'Age':[28,22]
})
# result = pd.concat([df1, df2],axis=0) #Concatenate: vertical / row Level / top on top
result = pd.concat([df1, df2], axis=1) ## Concatenate: horizontal / column Level / side on side

result
```

```python
# Merge in Pandas
# Performs join operations similar to relational databases like SQL JOINS

df1 = pd.DataFrame({
    'EmpID': [1, 2, 3],
    'Name': ['Ram', 'Ravi', 'Amit']
})
df2 = pd.DataFrame({
    'EmpID': [1, 2, 4],
    'Salary': [30000, 40000, 50000]

})
```

```python
pd.merge(df1,df2,on = 'EmpID')  #inner Join (default ) sirf matching data 
```

```python
#LEft Join Left table ka sab data + matching right 
pd.merge(df1,df2,on= ['EmpID'],how = 'left')
```

```python
#Right Join means right table ka sab data + matching data 
pd.merge(df1,df2,on = ['EmpID'],how = 'right')
```

```python
#Outer Join sab data dono tables ka 
pd.merge(df1,df2, on = ['EmpID'],how = 'outer')
```

```python
#Join type Result 
#inner only matching
#left left +matching right
#right  right + matching left 
#outer  All data 
```

```python
pd.merge(df1,df2, how = 'inner', left_on = 'EmpID',right_on='EmpID') # dono table mai name alag ho tab 
```
