```python
# Aggregation and GroupBy (Pandas) – Detailed Explanation

# Data Analysis me GroupBy + Aggregation bahut important concept hai. 
# Iska use hum data ko category wise group karke summary nikalne ke liye karte hain.
```

```python
# 🔷 1. GroupBy kya hota hai?

# 👉 groupby() ka matlab:

# “Same category wale data ko group karo”

# Example:

# Department wise salary
# City wise sales
# Gender wise marks

# df.groupby('column_name')
```

```python
import pandas as pd 
```

```python
data = {
    'Name': ['Ram', 'Ravi', 'Amit', 'Ram', 'Ravi'],
    'Dept': ['IT', 'HR', 'IT', 'HR', 'IT'],
    'Salary': [30000, 25000, 40000, 32000, 28000]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# 2. Aggregation kya hota hai?
# "Grouped data par calculation karna"

# Common functions:
# sum()
# mean()
# min()
# max()
# count()
```

```python
df['Dept'].value_counts()  # frequency of values in month column
```

```python
df[df['Dept']=='IT'].value_counts()  # frequency of values in month column where month = 1
```

```python
# Department  wise total salary  ['Salary'] Aggregation
df.groupby('Dept')['Salary'].sum()
```

```python
#Department wise Average Salary 
df.groupby('Dept')['Salary'].count()
```

```python
#count of employees per department 
df.groupby('Dept')['Name'].count()
```

```python
#Multiple aggregartions (Most Important)
df.groupby('Dept')['Salary'].agg(['sum','mean','min','max'])
```

```python
#Multiple column group by Dept + Name combination wise result 
df.groupby(['Dept','Name'])['Salary'].sum()
```

```python
#Named Aggregation (Advanced)
df.groupby('Dept').agg(
    total_salary = ('Salary','sum'),
    avg_salary = ('Salary','mean'),
    count_emp = ('Name','count')
)
```
