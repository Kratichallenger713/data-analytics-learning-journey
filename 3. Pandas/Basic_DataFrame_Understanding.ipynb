```python
import pandas as pd 
```

```python
data = {
    'Name': ['ram', 'ravi', 'ak', 'vj',"av","rj"],
    'Age': [26,37,28,19,4,3],
    'Salary': [10000, 3000, 40000, 50000,555,777]

}
```

```python
df = pd.DataFrame(data)
print(df)
```
