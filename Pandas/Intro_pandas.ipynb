```python
# Pandas ke Main Data Structures
# 1. Series
# 1D data structure

# Pandas Series

# Series Pandas ka 1-dimensional data structure hai.
# Ye ek single column ki tarah hota hai.

# Simple language me:

# 👉 List + Index = Series
```

```python
s =pd.Series([10,20,30,40])
print(s)
print(s[0])
```

```python
#Custom Index
s = pd.Series([100,200,300], index =['a','b','c'])
print(s)
print(s['b'])
print(s.dtype)
```

```python
#Dictionary se Series
data={
    'Name':'Ravi',
    'Course':'python',
    'Fees':5000
}
s =pd.Series(data)
print(s.values)
print(s.index)
print(s.dtype)
```

```python
# What is Dataframe?
# DataFrame Pandas ka ek 2-dimensional data structure hai jo data ko rows aur columns ke form mein 
#store karta hai,# bilkul Excel sheet ya database table ki tarah.

# DataFrame ek table hota hai jisme multiple rows aur columns hote hain.

# | Name   | Age | City    |
# | ------ | --- | ------- |
# | Aman   | 22  | Gwalior |
# | Rahul  | 24  | Indore  |
# | Simran | 21  | Bhopal  |
```

```python
df = pd.DataFrame([10,11,12],columns = ['Col_name']) #square brackets () parentheses
print(df)  #0 kyu kiya kyuki column ka name nahi diya hai isliye 0 de raha hai 
```

```python
print(type(df)) #check data type 
```
