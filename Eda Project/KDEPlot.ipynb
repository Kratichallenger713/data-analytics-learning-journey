```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# KDE Plot क्या है?

# मान लो आपके पास tax_amount है:
# 100
# 110
# 120
# 115
# 120
# 125
# 130
# 200
# 210
# 205
# आप जानना चाहते हो:

# Tax amount किस range के आसपास सबसे ज्यादा है?

# इसके लिए KDE Plot एक smooth curve बनाता है।
# Density
#   |
#   |          /\
#   |         /  \
#   |        /    \       /\
#   |_______/      \_____/  \____
#   |
#   +--------------------------------
#        100-130        200-210
#               Tax Amount
# यहाँ पहले वाला peak ज्यादा ऊँचा है।

# मतलब:

# 100–130 के आसपास tax amount की values ज्यादा concentrated हैं।

# sns.kdeplot(x=df['tax_amount'], fill=True)
# इसे बस ऐसे याद रखो:

# kdeplot() → data कहाँ ज्यादा concentrated है, उसकी smooth curve दिखाता है।

# और:

# fill=True

# का मतलब:

# Curve के नीचे का area fill कर दो।

# Histogram vs KDE

# इसे सबसे आसान तरीके से याद करो:

# Graph	क्या दिखाता है?
# Histogram	किस range में कितने records हैं
# KDE Plot	किस range में data ज्यादा concentrated है
# एक लाइन में exam answer

# अगर कोई पूछे KDE Plot क्या है?

# "KDE Plot numerical data की distribution को एक smooth curve के रूप में दिखाता है और इससे पता चलता है 
# कि data किस range में ज्यादा concentrated है
```

```python
df=pd.DataFrame({
    'customer':['A','B','C','D','E','F','G','H','I','J'],
    'tax_amount':[100,120,110,150,130,200,220,210,250,230]
})

plt.figure(figsize=(10,5))
sns.kdeplot(
   x=df['tax_amount'], #tax_amount के data की smooth distribution curve बनाओ। यानी हमें देखना है
    #कि tax amount किस range में ज्यादा concentrated है। curve आएगा।
    fill=True #curve के नीचे का area भी भर जाएगा। इसलिए graph को समझना आसान हो जाता है।
)
plt.title('Tax Amount Distribution')
plt.show()
```
