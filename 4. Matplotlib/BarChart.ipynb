```python
import matplotlib.pyplot as plt
```

```python
students = ['Ram','Shyam','Mohan']
marks = [80,90,70]
plt.bar(students,marks)
plt.show()
```

```python
#Title aur label 
plt.bar(students,marks)
plt.title('Student Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.show()
```

```python
#bar color 
plt.bar(students,marks,color='red')
plt.show()
```

```python
#Bar width 
plt.bar(
    students,
    marks,
    width=0.5
)
plt.show()
```

```python
#Grid Add Karna 
plt.bar(students,marks)
plt.grid(axis='y') # bar charts me mostly axis ='y' use hoti hai 
plt.show()
```

```python
#Data Labels (important)
students = ['Ram','Shyam','Mohan']
marks = [80,90,70]
bars = plt.bar(students,marks)
plt.bar_label(bars,padding=3) # value automatically bar ke upar 
plt.show()
```

```python
#horizontal Bar chart 
plt.barh(students,marks) # horizontal bar chart
plt.show()
```

```python
#Multiple color 
colors = ['red' , 'green', 'blue']
plt.bar(
    students,
    marks,
    color = colors
)
plt.show()
```

```python
#Figure size 
plt.figure(figsize=(8,5))
plt.bar(students,marks)
plt.show()
```

```python
# Grouped Bar Chart
# Grouped Bar Chart ka use ek hi category ke andar multiple data sets ko compare karne ke
#Liye kiya jhta hai,
# jaise Students ke Maths aur Science marks.
import numpy as np

students = ['Ram','Shyam','Mohan']
maths = [80,90,70]
science = [75,85,65]

x = np.arange(len(students))
plt.bar(x-0.2, maths, width=0.4 , label='Maths')
plt.bar(x+0.2, science,width=0.4,label='Science')

plt.xticks(x,students)
plt.legend()
plt.show()
#Ek student ke liye 2 subjects compare kar sakte ho 
# x = [0,1,2]-> categories ki positions
#x -0.2 -> left shift
# x+0.2 + right shift
# width=0.4 + bar ki width
# plt.xticks() - numbers ko names me badalta hai
# plt. Legend() + Maths aur Science ka Label dikhata hai
```

```python
#Stacked barchart 
students = ['Ram','Shyam','Mohan']
maths = [80,90,70]
science = [75,85,65]

plt.bar(students,maths,label='Maths')
plt.bar(students,science, label='Science')
plt.legend()
plt.show()
# yaha bars ek dusre par stack hoti h 
```

```python
#pandas ke sath 
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Student':['Ram','Shyam','Mohan'],
    'Marks':[80,90,70]
})
df.plot(kind ='bar',x='Student',y='Marks')
plt.show()
```

```python
months = ['Jan','Feb','Mar','Apr','May']
sales = [10000,15000,18000,25000,35000]

bars = plt.bar(months,sales)
plt.bar_label(bars)
plt.title("Monthly Sales Report")
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
          
```
