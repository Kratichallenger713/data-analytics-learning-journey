```python
# zip() Python ka built-in function hai jo 2 ya usse zyada lists ko pair (jodi) bana kar ek saath iterate karne ke liye use hota hai.
```

```python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [10,20,25,35,45]

plt.plot(months, sales, marker='o')

for x, y in zip(months, sales):
    print(x,y)  #Yahi values plt.text() graph par likh deta hai.
    plt.text(x, y, str(y))

plt.show()
```

```python
# Grid Add Karna
plt.plot(months,sales)

plt.grid(True) #axis='y'x linestyle='--' linestyle=':' color='red'   linewidth=2 alpha=0.5 0 = invisible
# 1 = fully visible

plt.show()
# Grid graph ko read karne aur values compare karne ko easy banati hai.
```
