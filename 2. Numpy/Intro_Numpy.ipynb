#Create array

```python
list1 = [1,2,3]
print(list1)
```

```python
arr_list1 = np.array(list1) # list is slower than array that's why we convert list into array
print(arr_list1)
```

```python
print(type(arr_list1)) #N-dimensional array (multi-dimensional array)
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
arr_list2 = np.array([list1,list2])
print(arr_list2)
```

```python
#axis 0-> rows 
#axis 1->column
#2 rows
# 3 column
#arr[0[0] = 1
#arr[0][1] = 6
# first index = row
# second index = column
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr_list3 = np.array([[list1,list2,list3]])
print(arr_list3)
```

```python
print(type(arr_list3))
```

```python
print(" list 3 ->",arr_list3.ndim)
print("list 1 ->",arr_list1.ndim)
print("list 2->",arr_list2.ndim)
```
