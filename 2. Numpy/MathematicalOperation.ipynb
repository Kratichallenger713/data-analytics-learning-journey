```python
#Statisctical functions
a = np.array([[1,2,3], [4,5,6]])
print(a)
```

```python
print(np.sum(a))  #sum of all elements of an array
```

```python
print(np.mean(a)) #average
print(np.median(a)) #middle value
print(np.std(a)) #standard deviation (Spread)
print(np.min(a)) #smallest
print(np.max(a))#Largest
```

```python
sales =np.array([5000,7000,6500,8000, 9000, 7500])
print("Total sales",np.sum(sales))
print("Highest sales",np.max(sales))
print("Lowest sales",np.min(sales))
print("Average sales", np.mean(sales))
print("SD sales",np.std(sales))
```
