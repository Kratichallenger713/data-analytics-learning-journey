```python
import numpy as np
```

```python
#Reshape Array : Array  ke data ko same rakhte hue uska  structure  change karna 
#Flatten Array : 2D ya 3 D array ko ek single line(1 D) me convert karna
```

```python
arr1 = np.array([1,2,3,4,5,6])  #shape (1,6) -> 1 rows and 6 columns 
print(arr1)
```

```python
reshaped =arr1.reshape((6,1))  #shape = (6,1 ) -> 6 rows and 1 columns
print(reshaped)
```

```python
reshaped2 = arr1.reshape((2,3))  # reshape array from (6,1) to  (2,3) -> 2 rows & 3 coulmns
print(reshaped2)
```

```python
reshaped3 = arr1.reshape((3,2))  # reshape array from (2,3) to  (3,2) -> 3 rows & 2 coulmns
print(reshaped3)
```

```python
print(reshaped2.flatten())  #flattened array from higher  2D dimensional (2,3) to 1 Dimensional
```

```python
#Array Stacking : Do Ya zyada arrays ko jodna ( combine karna ) type 3 (vertical stack np.vstack((a,b)))
#Horizontal Stack : np.hstack((a,b))
#type 2 (Vertical Stack np.vstack((a,b)))
# Horizontal Stack: np.hstack((a,b))
# Array Splitting: Ek array ko chhote parts me todna.
```

```python
# stacking
a = np.array([1,2,3])
b = np. array([4,5,6])
stacked = np.stack((a,b))
print(stacked)

print(np.vstack((a,b))) # verticat stacking - row wise -> output array is 2D
print(np.hstack((a,b))) # horizontal stacking - column wise -> output array is 1D
```

```python
#Splitting 
c = np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3) # horizontal split  -> input and output array are same dimension
print(hsplit)
for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2) # vertical split -> input & output array are same dimension
for s in vsplit:
    print(s)
```
