```python
# Seaborn is a Python library used for data visualization. It is built on top of Matplotlib and provides simpler functions for creating attractive statistical graphs.
# In simple words:
# Seaborn = easier and more statistical-focused data visualization in Python.
# # seaborn python
# Seaborn ek Python Visualization Library hai jo Matplotlib ke upar bani hai. Iska use attractive aur 
# professional graphs banane ke liye kiya jata hai.Data Analytics aur Machine Learning 
#projects me Seaborn bahut popular hai.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 60, 80, 100]
}

df = pd.DataFrame(data)

sns.scatterplot(x="Hours", y="Marks", data=df)

plt.show()
```

```python
data={
    "Ad_Budget":[10,15,20,25,30,35,40],
    "Sales":[50,65,80,95,120,130,150]
}
df=pd.DataFrame(data)
sns.scatterplot(x="Ad_Budget",y="Sales",data=df)
plt.title("Advertising Budget vs Sales")
plt.show()
```

```python
# 5. Box Plot
# Minimum, Maximum, Median aur Outliers dikhata hai.
# Box Plot thoda advanced graph hai. Iska use data ka spread, median, aur outliers dekhne ke liye hota hai.
```

```python
marks=[20,25,30,35,40,45,50,55,60,95]
sns.boxplot(y=marks)
plt.title("Student Marks")
plt.show()
```

```python
# Graph Ko Kaise Samjhen?

# Data:

# 20, 25, 30, 35, 40, 45, 50, 55, 60, 95
# Median (Middle Value)
# 40 aur 45 ke beech
# ≈ 42.5

# Box ke beech wali line = Median

# Box
# |--------|
# |        |
# |--------|

# Box ke andar most data hota hai.

# Yahan lagbhag:

# 30 se 55

# ke beech.

# Outlier
# 95

# Baaki sabse bahut door hai.

# Graph me ye alag dot (●) ki tarah dikhega.

# Real Life Meaning

# Class ke sab students ke marks:

# 20 se 60

# ke beech hain.

# Lekin ek student ne:

# 95

# score kiya.

# To 95 Outlier hai.

# Interview Line

# Box Plot data ka spread, median aur outliers dikhata hai. Is example me 95 ek outlier hai kyunki ye baaki values 
#se bahut zyada hai.

# Yaad Rakho
# Box  = Most Data
# Line = Median
# Dot  = Outlier

# Ye Box Plot samajhne ka sabse simple rule hai
```

```python
salary = [10000,12000,15000,18000,20000,22000,25000,30000,100000]
sns.boxplot(y=salary)
plt.title("Employee Salary")
plt.show()
```

```python
# Data Ko Samjho
# 10000
# 12000
# 15000
# 18000
# 20000
# 22000
# 25000
# 28000
# 30000
# 100000

# Yahan:

# Most employees ki salary = ₹10,000 se ₹30,000
# Ek employee ki salary = ₹1,00,000
# Box Plot Kya Dikhayega?
#         ●  ← 100000 (Outlier)

#       |
#   ---------
#  |         |
#  |   BOX   |
#  |         |
#   ---------
#       |
# Meaning

# ✅ Box = Most employees ki salary range

# ✅ Middle line = Median Salary

# ✅ Upar wala dot (100000) = Outlier

# Real Life Meaning

# Company me 9 employees ki salary normal hai:

# 10000 - 30000

# Lekin manager ki salary:

# 100000

# Hai.

# Isliye manager ki salary Outlier ban jayegi.

# Interview Answer

# "Box Plot salary distribution ko show karta hai. Is example me ₹1,00,000 salary outlier hai kyunki ye baaki salaries se bahut zyada hai."

# Box Plot Yaad Rakhne Ki Trick
# Box  = Normal Data
# Line = Median
# Dot  = Outlier

# Student Marks aur Employee Salary dono Box Plot ke bahut common Data Analytics examples hain.
```

```python
tips=sns.load_dataset("tips")
print(tips.head())
sns.boxplot(x="day",y="total_bill",data=tips)
plt.show()
```

```python
#countplot 
#category ki counting karta hai 
sns.countplot(x="day",data=tips)
plt.show()
```

```python
#Heatmap 
#it shows the correlation between the data 
data=[
    [10,20,30],
    [40,50,60],
    [70,80,90]
]
sns.heatmap(data,annot=True,cmap="Greens") #read,Greens,coolwarm,virdis,Blues
plt.show()
#jis value ki color dark hogi -> value zyada hogi
# jis value ki color light hogi -> value kam hogi
```

```python
data ={
    "Math":[80,70,90],
    "Science":[75,88,92],
    "English":[85,79,91]
}
df = pd.DataFrame(data)
sns.heatmap(df,annot=True,cmap="Greens")
plt.show()
```

```python
data={
    "Math":[85,90,75,60,95],
    "Science":[80,88,70,65,92],
    "English":[78,85,72,68,90]
}
df=pd.DataFrame(data)
sns.heatmap(df,annot=True,cmap="Blues",linewidth=2,linecolor="red",fmt=".0f")
plt.title("student Marks")
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.show()
```
